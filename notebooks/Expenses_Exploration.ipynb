## c. Is the dataset complete?  no missing values?

```python
missing_values = df.isnull().sum()
print(missing_values)

total_missing = missing_values.sum()
if total_missing == 0:
    print("\nThe dataset is complete: no missing values found.")
else:
    print(f"\nThe dataset is NOT complete: {total_missing} missing values found across the columns above.")
```
